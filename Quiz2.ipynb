3. Linda bought a graduation suit for XAF 45000 and a pair of shoe for XAF 60000. She gave $5000 to the shopkeeper. How much money did the shopkeeper return to her?

```python
import requests as req
import json as json

usd_amount= 5000 # Amount in Dollat($)
suit_cost= 45000 # Amount in XAF
shoe_cost= 60000 # Amount in XAF

 # Converting the Dollar($) to XAF using Exchange Rate
def get_exchange_rate(base_currency, target_currency):
  #Fetch exchange rate from Exchange_Rate API
    url= f"https://api.exchangerate-api.com/v4/latest/{base_currency}"
    try:
        response= req.get(url)
        data= response.json()
        
        if target_currency in data['rates']:
            rate= data['rates'][target_currency]
            return rate
        else:
             print(f"Currency {target_currency} not found")
             return None
    except Exception as e:
        print(f"Error fetching exchange rate: {e}")
        return None

  # calculations      

exchange_rate= get_exchange_rate('USD','XAF')
xaf_amount= usd_amount * exchange_rate
balance= xaf_amount - (suit_cost + shoe_cost)
[suit_cost, shoe_cost, balance]

#outputs
for asset_name, asset_value in [("Suit Cost", suit_cost),("Shoe cost", shoe_cost), ("Balance", balance)]:
    print(f"{asset_name}: {asset_value: ,.2f} XAF")


```

9. There are 24 classrooms in a LMUI. 986 students study there. what is the Equal number of students sits in each room?

```python
# Given information
total_classrooms = 24
total_students = 986

# Calculate students per classroom
students_per_classroom = total_students / total_classrooms

# Calculate for equal distribution (whole numbers)
students_per_classroom_whole = total_students // total_classrooms  # Integer division
remaining_students = total_students % total_classrooms  # Remainder

# Display results
print("=" * 50)
print("LMUI CLASSROOM STUDENT DISTRIBUTION")
print("=" * 50)
print(f"Total number of classrooms: {total_classrooms}")
print(f"Total number of students: {total_students:,}")
print(f"\nExact division: {students_per_classroom:.2f} students per classroom")
print("=" * 50)

# Equal distribution with whole numbers
print(f"\nEqual distribution (whole numbers):")
print(f"  Students per classroom: {students_per_classroom_whole}")
print(f"  Remaining students: {remaining_students}")
print(f"\nDistribution plan:")
print(f"  {total_classrooms - remaining_students} classrooms with {students_per_classroom_whole} students each")
print(f"  {remaining_students} classrooms with {students_per_classroom_whole + 1} students each")

# Verification
verification = (students_per_classroom_whole * (total_classrooms - remaining_students)) + ((students_per_classroom_whole + 1) * remaining_students)
print(f"\nVerification: {verification} students = {total_students} students ")
```

10. 986 students of LMUI want to go to SEME beach by bus. If one bus can carry 75 students, how many buses are required to carry all the students?

```python
import math
# Given information
total_students = 986
students_per_bus = 75

# Calculate number of buses required
buses_needed_exact = total_students / students_per_bus
buses_needed = math.ceil(total_students / students_per_bus)  # Ceiling division (round up)


# Calculate students in last bus
students_in_last_bus = total_students % students_per_bus
if students_in_last_bus == 0:
    students_in_last_bus = students_per_bus

# Display results
print("=" * 50)
print("LMUI TO SEME BEACH - BUS TRANSPORTATION")
print("=" * 50)
print(f"Total students going to SEME beach: {total_students:,}")
print(f"Capacity per bus: {students_per_bus} students")
print(f"\nExact calculation: {buses_needed_exact:.2f} buses")
print(f"Buses required: {buses_needed} buses")
print("=" * 50)

# Distribution breakdown
print(f"\nBus distribution:")
print(f"  Buses 1-{buses_needed - 1}: {students_per_bus} students each (Full)")
print(f"  Bus {buses_needed}: {students_in_last_bus} students (Last bus)")

# Verification
total_capacity = buses_needed * students_per_bus
print(f"\nVerification:")
print(f"  Total bus capacity: {buses_needed} × {students_per_bus} = {total_capacity} seats")
print(f"  Students to transport: {total_students}")
print(f"  Empty seats: {total_capacity - total_students}")
```

13. Maria bought 96 toys priced equally for XAF 12960. The amount of XAF 1015 is still left with her. Find the cost of each toy and the amount she had.

```python
# Given information
number_of_toys = 96
total_cost_of_toys = 12960  # XAF
amount_left = 1015  # XAF

# Calculate cost per toy
cost_per_toy = total_cost_of_toys / number_of_toys

# Calculate total amount Maria had
total_amount = total_cost_of_toys + amount_left

# Display results
print("=" * 50)
print("MARIA'S TOY PURCHASE ANALYSIS")
print("=" * 50)
print(f"Number of toys bought: {number_of_toys}")
print(f"Total cost of toys: XAF {total_cost_of_toys:,}")
print(f"Amount left after purchase: XAF {amount_left:,}")
print("=" * 50)

# Main answers
print(f"\nAnswers:")
print(f"  1. Cost of each toy: XAF {cost_per_toy:,.2f}")
print(f"  2. Total amount Maria had: XAF {total_amount:,}")
print("=" * 50)

# Calculation breakdown
print(f"\nCalculation breakdown:")
print(f"  Cost per toy = {total_cost_of_toys:,} ÷ {number_of_toys} = XAF {cost_per_toy:,.2f}")
print(f"  Total amount = {total_cost_of_toys:,} + {amount_left:,} = XAF {total_amount:,}")

# Verification
verification = (cost_per_toy * number_of_toys) + amount_left
print(f"\nVerification:")
print(f"  ({cost_per_toy:,.2f} × {number_of_toys}) + {amount_left:,} = XAF {verification:,.2f}")
print(f"  Total matches: XAF {total_amount:,} ")
```

> **EXERCISE**: CamairCo wants to fly a plane to the from the Douala International airport to The Nsimalen Airport in Yaounde. Flying the plane costs XAF120000 . So far, 29 people have signed up for the trip. If the CamairCo charges XAF20000 per ticket, what is the profit made by the company? Create variables for each numeric quantity and use appropriate arithmetic operations.

```python
##Define Variables
flight_cost = 120000  # XAF - Cost to operate the plane
number_of_passengers = 29  # Number of people who signed up
ticket_price = 20000  # XAF - Price per ticket
```

```python

# Calculate total revenue from ticket sales
total_revenue = number_of_passengers * ticket_price

# Calculate profit (Revenue - Cost)
profit = total_revenue - flight_cost

# Display results
print("=" * 60)
print("CAMAIRCO FLIGHT PROFIT ANALYSIS")
print("Route: Douala International Airport → Nsimalen Airport (Yaounde)")
print("=" * 60)

# Show all variables
print(f"\nFlight Details:")
print(f"  Operating cost: XAF {flight_cost:,}")
print(f"  Ticket price: XAF {ticket_price:,}")
print(f"  Number of passengers: {number_of_passengers}")

# Show calculations
print(f"\nFinancial Breakdown:")
print(f"  Total revenue: {number_of_passengers} × XAF {ticket_price:,} = XAF {total_revenue:,}")
print(f"  Operating cost: XAF {flight_cost:,}")
print(f"  Profit: XAF {total_revenue:,} - XAF {flight_cost:,} = XAF {profit:,}")

print("=" * 60)

# Final answer
if profit > 0:
    print(f"\n✓ PROFIT: XAF {profit:,}")
    print(f"  Profit margin: {(profit/total_revenue)*100:.2f}%")
elif profit < 0:
    print(f"\n✗ LOSS: XAF {abs(profit):,}")
else:
    print(f"\n= BREAK-EVEN: XAF 0")

```
