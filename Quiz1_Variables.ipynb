**SENARIO:**
>3. James works 28 days a month as a junior data analyste for Micro1 and earns $10 per hour..If James worked for 6 months how much did he earn?

```python
# data collection
initial_work_days= 28
revenue_earned_per_hour= 10
new_work_duration= 6  # in months

# Calculations
# We want to obtain the revenue James would accumulate for working at a duration of 6 months
# Convereting months - days -hours
current_work_hours= (initial_work_days * 24)
current_monthly_revenue= (current_work_hours * revenue_earned_per_hour)
new_work_hours= (new_work_duration * initial_work_days * 24)
new_revenue= (new_work_hours * revenue_earned_per_hour)

# Outputs
print(f"Current Revenue for {initial_work_days} days: ${current_monthly_revenue}\nNew Revenue for {new_work_duration} months: ${new_revenue}")
```
